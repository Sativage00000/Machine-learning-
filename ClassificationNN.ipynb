```python
import pandas as pd
import numpy as np
import seaborn as sb
import tensorflow as tf
```

```python
dataset = pd.read_csv("breast_cancer.csv")
```

```python
X = dataset.iloc[ : , 3:]
Y = dataset.iloc[ : , 2]
```

```python
dataset
```

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler
mmScaler = MinMaxScaler()
X = mmScaler.fit_transform(X.astype(float))
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, Y, test_size=0.1, random_state=6)
```

```python
from tensorflow.keras.optimizers import Adam
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Activation
from tensorflow.keras.optimizers import Adam
```

```python
import pandas as pd
import numpy as np
import seaborn as sb
import tensorflow as tf
```

```python
model = Sequential() 
model.add(Dense(15, activation="relu")) 
model.add(Dense(15, activation="relu")) 
model.add(Dense(15, activation="relu")) 
model.add(Dense(2, activation=tf.nn.softmax))

model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])

modelMetrics = model.fit(X_train, y_train.values, epochs=200)
```

```python
val_loss, val_acc = model.evaluate(X_test, y_test.values)
```

```python
predictions = model.predict(X_test)
```

```python
print(predictions)
```

```python
for i in range (0, len(X_test)):
    print("Predicted -> ", np.argmax(predictions[i]), " Actual -> ", y_test.iloc[i])
```

```python
from pandas import DataFrame
from sklearn.metrics import confusion_matrix
from sklearn.metrics import classification_report

predictedList = []
for i in range (0, len( X_test)):
    predictedList.append(np.argmax(predictions[i]))
predictedList
```

```python
predictions = DataFrame(predictedList)
```

```python
from sklearn.metrics import accuracy_score
print(classification_report(y_test, predictions))
print(confusion_matrix(y_test, predictions))
accuracy = accuracy_score(y_test, predictions)
print(accuracy)
```
